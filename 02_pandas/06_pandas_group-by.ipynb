```python
import pandas as pd
data=pd.read_excel("ESD.xlsx")
print(data)
```

```python
#use of group by
gb=data.groupby("Department").agg({"Gender":"count"})  
print(gb)

gb1=data.groupby(["Department","Gender"]).agg({"Annual Salary":"mean"})
print(gb1)

gb2=data.groupby(["Department","Gender"]).agg({"Annual Salary":"max"})
print(gb2)

gb3=data.groupby(["Department","Gender"]).agg({"Annual Salary":"max","Age":"min"})
print(gb3)
```
