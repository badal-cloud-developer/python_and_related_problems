```python
import pandas as pd
data1 = {"emp id":["E01","E02","E03","E04","E05","E06"],
       "names":["badal","ram","oli","bbsh","ashish","purne"],
       "age":[20,30,40,35,80,25]}
data2 = {"emp id":["E01","E02","E03","E04","E05","E06"],
    "salary":[200000,30000,40000,35000,80000,25000]}

df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)
print(df1)
print()
print(df2)
```

```python
print(pd.merge(df1,df2,on="emp id")) #emp id same hunu paryo merge huna.like innerjoin

```

```python
data1 = {"emp id":["E01","E02","E03","E04","E05","E06"],
       "names":["badal","ram","oli","bbsh","ashish","purne"],
       "age":[20,30,40,35,80,25]}
data2 = {"emp id":["E01","E07","E03","E04","E09","E06"],
    "salary":[200000,30000,40000,35000,80000,25000]}

df11 = pd.DataFrame(data1)
df22 = pd.DataFrame(data2)

print(pd.merge(left=df11,right=df22,on="emp id",how="left")) #left join,left ko sabai ani right ko common matra
```

```python
#concatinate

data1 = {"emp id":["E01","E02","E03","E04","E05","E06"],
       "names":["badal","ram","oli","bbsh","ashish","purne"],
       "age":[20,30,40,35,80,25]}
data2 = {"emp id":["E07","E08","E09","E10","E11","E12"],
    "names":["sita","gita","rita","sarita","nirtuta","pabitra"]}
df1=pd.DataFrame(data1)
df2=pd.DataFrame(data2)
print(pd.concat([df1,df2]))
```
