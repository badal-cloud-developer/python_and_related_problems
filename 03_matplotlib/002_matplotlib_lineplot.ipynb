```python
import matplotlib.pyplot as plt

x = ["day1","day2","day3","day4","day5"]
y= [300,400,200,250,400]
y1=[400,450,300,500,600]

plt.plot(x,y,marker="o",ls= "--",color="red",label= "week1")
plt.plot(x,y1,marker="o",ls= "--",color="yellow",label="week2")
plt.legend()
plt.show()
```
