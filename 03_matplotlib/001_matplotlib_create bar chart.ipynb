```python
import matplotlib.pyplot as plt

y = [70,80,69,90,95]
x = ["part1", "part2", "part3" , "part4" , "part5"]

colors = ["red","blue","green","yellow","orange"]
plt.bar(x , y,color=colors, edgecolor = "black") #creates bar chart 

plt.xlabel("diff parts of harry potter")
plt.ylabel("popularity")
plt.title("popularity of different parts of harry potter",fontsize= 15)

plt.show()  #displays bar chart from above data
```

```python
#displaying bar chart from the .csv or .xlsx file

import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_excel("expense3.xlsx")
df = pd.DataFrame(data)

# Clean missing values before plotting
df = df.dropna(subset=["Payment Mode", "Amount"])

print(df)
plt.bar(df["Payment Mode"], df["Amount"])
plt.show()

```
