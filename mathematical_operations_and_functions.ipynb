```python
#array addition

import numpy as np
arr1=np.array([10,20,30,40])
arr2=np.array([50,60,70,80])
print(arr1+arr2)
print(np.add(arr1,arr2))

arr1=np.array([[10,20],[30,40]])
arr2=np.array([[50,60],[70,80]])
print(arr1+arr2)
```

```python
#array subtraction

import numpy as np
arr1=np.array([10,20,30,40])
arr2=np.array([50,60,70,80])
print(arr1-arr2)
print(np.subtract(arr1,arr2))

arr1=np.array([[10,20],[30,40]])
arr2=np.array([[50,60],[70,80]])
print(arr1-arr2)

```
