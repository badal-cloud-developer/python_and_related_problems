```python
import numpy as np
import statistics as stats

baked_food=[200,300,150,130,200,280,170,100]  #lets say hunuparne carbohydrade 200 ho.
a=np.array(baked_food)

print(np.mean(baked_food))  #gives average value

print(np.median(baked_food)) #gives central value after sorting

print(stats.mode(baked_food)) #gives most recurring value
```

```python
print(np.std(baked_food)) #aru value haru mean bata kati tada tada xan vanne dinxa
                          #if low std means means bata najikai ka xan
                          #else need quality control
```

```python
print(np.var(baked_food))  #it is the square of standard deviation
```
